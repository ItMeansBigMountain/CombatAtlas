# Square each value in array
[ 1 , 3 , 3 , 7 ] ==> [ 1 , 9 , 9 , 49 ]

```python
arr = [ 1 , 3 , 3 , 7 ]

for x in range(  0  ,  len(arr) ,  1):
    arr[x] = arr[x] * arr[x]

arr
```

# Return array count greater than VAR

```python
arr = [ 1 , 3 , 3 , 7 ]
var = 5
count = 0

for x in range( 0  , len(arr) , 1  ):
    if arr[x] > var:
        count += 1




print( count )
```

# Change all negative values to zero
[ 1 , -3 , -3 , 7 ]

```python
arr = [ 1 , -3 , -3 , 7 ]

for x in range( 0  , len(arr) , 1  ):
    if arr[x] < 0:
        arr[x] = 0

arr
```

# Shift array values to the left
Leave the last value zero

[ 1 , 3 , 3 , 7 ] ==> [ 3 , 3 , 7 , 0 ]

```python
arr = [ 1 , 3 , 3 , 7 ]

for x in range( 0  , len(arr)-1 , 1  ):
    arr[x] = arr[x+1]
arr[ len(arr) - 1 ] = 0

arr
```
